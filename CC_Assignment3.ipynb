```python
#Ques 6
df=pd.read_csv('employees.csv')
#(a)
print("The shape of the employees csv is")
print(df.shape)

#(b)
print("The summary of the employeeds csv is")
print(df.info)

#(c)
print("The descriptive statistics of the employees csv is")
print(df.describe())

#(d)
print(df.head())
print(df.tail(3))

#(e)
print("Average salary is",df['Salary'].mean())
print("The total bonus paid is",df['Bonus'].sum())
print("The age of the youngest employee is",df['Age'].min())
print("The highest performance rating is",df['Rating'].max())

#(f)
print("The sorted salary order is",df.sort_values(by='Salary',ascending=False))

#(g)
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"
df["Performance_Category"] = df["Rating"].apply(performance_category)
display(df)

#(h)
print("The missing values are:")
print(df.isnull().sum())

#(i)
df.rename(columns={'Employee_ID':'ID'},inplace=True)

#(j)
print("Employees wth more than 5 years of experience:")
print(df[df['Years_of_Experience']>5])
print(df[df["Department"]=='IT'])

#(k)
df['Tax']=df['Salary']*0.10

#(l)
output_file = "modified_employee_dataset.csv"
df.to_csv(output_file, index=False)
files.download(output_file)
```
