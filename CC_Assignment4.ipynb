```python
#Ques1
import pandas as pd
roll_no="42"
fixed_entries = [
{"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
"keywords": "fee cost price charge", "category": "billing"},
{"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
"keywords": "password reset login", "category": "account"},
{"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
"keywords": "hours timing open time", "category": "general"},
{"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
"keywords": "pay payment upi fee", "category": "billing"},
]
digit1=int(roll_no[0])
digit2=int(roll_no[1])

categories=['billing','account','general']
category1=categories[digit1%3]
category2=categories[digit2%3]

my_entries=[
    {"question":"how do I update my registered mobile number",
     "answer":"Go to the setting option and update the mobile number.",
     "keywords":"mobile number update phone",
     "category":category1
     },
    {
        "question": "where can I find the frequently asked questions",
        "answer": "You can find FAQs in the Help or Support section.",
        "keywords": "faq help support",
        "category": category2
    }
]
all_entries = fixed_entries+my_entries
df=pd.DataFrame(all_entries)
print(df)
```

```python
#Ques2
def score_query(query,df):
  query_words=set(query.lower().split())
  results=[]
  for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        matches = query_words.intersection(keywords.union(question_words))
        score = len(matches)
        if score > 0:
            results.append((index, score))
  results.sort(key=lambda x: x[1], reverse=True)
  return results

query = input("Enter your query: ")

results = score_query(query, df)

print("Matching entries ranked by confidence:")

if results:
    for index, score in results:
        print("Score:", score)
        print(df.iloc[index])
        print()
else:
    print("No matching FAQ found.")
```

```python
#Ques3
def same_category(category_name,df):
  return df[df["category"] == category_name][
        ["question", "answer", "keywords", "category"]
    ]

print("Entries in category",category1)
print(same_category(category1,df))
```

```python
#Ques 6
def score_query_with_ties(query, df):
    query_words = set(query.lower().split())
    scores = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        matches = query_words.intersection(keywords.union(question_words))
        score = len(matches)

        scores.append((index, score))

    highest_score = max(score for index, score in scores)

    best_matches = [
        (index, score)
        for index, score in scores
        if score == highest_score and score > 0
    ]
    if best_matches:
        print("\nHighest score:", highest_score)

        if len(best_matches) > 1:
            print("Tie detected! Multiple matching entries:")
        else:
            print("Single best match:")

        for index, score in best_matches:
            print("\nQuestion:", df.iloc[index]["question"])
            print("Answer:", df.iloc[index]["answer"])
            print("Category:", df.iloc[index]["category"])
            print("Score:", score)

    else:
        print("No matching FAQ found.")


# Demonstration of tie
print("\nQuery producing a tie")
score_query_with_ties("fee", df)

# Demonstration without tie
print("\nQuery without a tie")
score_query_with_ties("password", df)
```
