```python
import numpy as np
#Ques1
arr=np.array([1,2,3,4,5])
print("Original array ",arr)
print("Addition of 2 in array",arr+2)
print("Multiplication of array with 3",arr*3)
print("Division of array by 2",arr/2)
```

```python
#Ques2
#(a)
arr = np.array([1, 2, 3, 6, 4, 5])
print("The reverse of the array is",arr[::-1])
#(b)
x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
print("The most frequent value in the array x is",np.bincount(x).argmax())
print("The most frequent value in the array y is",np.bincount(y).argmax())
```

```python
#Ques4
Dhiren = np.linspace(10,100,25)
print("Original array Dhiren",Dhiren)
print("Dimension of Dhiren",Dhiren.ndim)
print("Shape of Dhiren array",Dhiren.shape)
print("Total number of elements in Dhiren array",Dhiren.size)
print("The data types of elemts in the Dhiren array",Dhiren.dtype)
print("The total number of bytes consumed is",Dhiren.nbytes)

transpose = Dhiren.reshape(1, 25).T
print("Transpose using reshape():")
print(transpose)

print("Transpose using T:")
print(Dhiren.T)
```

```python
#Ques5
ucs420_dhiren=np.array([[10,20,30,40],[50,60,70,80],[90,15,20,35]])
print("Mean",np.mean(ucs420_dhiren))
print("Median",np.median(ucs420_dhiren))
print("Max",np.max(ucs420_dhiren))
print("Min",np.min(ucs420_dhiren))
print("UNique elements",np.unique(ucs420_dhiren))

reshaped_ucs420_dhiren=ucs420_dhiren.reshape(4,3)
print("Reshaped array",reshaped_ucs420_dhiren)

resized_ucs420_dhiren=np.resize(ucs420_dhiren,(2,3))
print("Resized array",resized_ucs420_dhiren)
```
