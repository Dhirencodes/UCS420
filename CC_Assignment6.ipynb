```python
import numpy as np
#Ques1
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temperature=temperature+2
print(temperature)

temp_f = ((temperature*9)/5) + 32
print("Temperature in fahrenheit is")
print(temp_f)

print("Temperature greater than 32",temperature[temperature>32])

print("The count of values that exceed 32 are",np.sum(temperature>32))

#Vectorization allows an operation to be performed on the entire NumPy array at once instead of processing each value using a loop
```

```python
#Ques2
steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])

print("The total number of steps are",steps.sum())
print("The mean of the steps is",steps.mean())
print("The max of the steps is",steps.max())
print("The min of the steps is",steps.min())

print("Total Steps for Each Day:")
print(np.sum(steps, axis=0))

print("Total Steps for Each User:")
print(np.sum(steps, axis=1))

max_index = np.argmax(steps)
position = np.unravel_index(max_index, steps.shape)
print("Index returned by argmax:", max_index)
print("Position of Maximum:", position)
```

```python
#Ques3
original = np.array([1, 2, 3, 4, 5, 6])
subset = original[1:5]
print(subset)

subset[0]=999
print("Original",original)
print("Subset",subset)

copy_arr = original[1:5].copy()
copy_arr[0]=500
print("Original",original)
print("Subset",subset)
print("Copy",copy_arr)

matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)

print("First Row:")
print(matrix[0])

print("Last Row:")
print(matrix[-1])

print("Second Column:")
print(matrix[:, 1])

print("Selected Elements:")
print(matrix[1:3, 2:4])

flattened = matrix.flatten()
print("Flattened Matrix:")
print(flattened)

raveled=matrix.ravel()
print("Raveled matrix:")
print(raveled)

raveled[0]=111
print("Raveled matrix",raveled)
print("Original matrix",original)

flattened[0]=222
print("Flattened matrix",flattened)
print("Original matrix",original)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data Type:", matrix.dtype)
```

```python
#Ques4
X = np.array([
[6, 70, 3],
[5, 50, 6],
[8, 80, 2],
[4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("Shape:", X.shape)
print("Dimensions:", X.ndim)

XT = X.T
print("Transpose:",XT)
#After transposing, each row represents one feature across all users

XTX = X@XT
print("Matrix product:",XTX)

inverse = np.linalg.inv(XTX)
print("Inverse matrix:",inverse)

beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Beta:")
print(beta)
#The three coefficients are basically the coefficiemts of sleep, activity and stress
#where sleep's is -0.15921454, activity is 0.14083853 and stress is 10.06302242

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted Assistance Score:", predicted_score)
```
